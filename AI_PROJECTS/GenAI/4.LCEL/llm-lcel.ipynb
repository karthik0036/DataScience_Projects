```python
import os
from dotenv import load_dotenv

load_dotenv()

groq_api_key = os.getenv("GROQ_API_KEY")
```

```python
from langchain_groq import ChatGroq 

model = ChatGroq(model="llama-3.1-8b-instant",groq_api_key=groq_api_key)
model
```

```python
from langchain_core.messages import HumanMessage,SystemMessage

messages = [
    SystemMessage(content="Translate the texts from english to french"),
    HumanMessage("hello , how are you doing?Thanks")
]

response = model.invoke(messages)
response
```

```python
from langchain_core.output_parsers import StrOutputParser

parser = StrOutputParser()
parser.invoke(response)
```

```python
##using LCEL - chain the components
chain = model | parser
chain.invoke(messages)
```

```python
#using prompt template 

from langchain_core.prompts import ChatPromptTemplate

generic_template = "Translate the text into given {language}"

prompt = ChatPromptTemplate.from_messages(
    [
        
        ("system",generic_template),("user","{text}")
    ]
)
```

```python
result = prompt.invoke({"language":"French","text":"Hello,How are you"})
result
```

```python
result.to_messages()
```

```python
chain = prompt | model |parser
chain.invoke({"language":"German","text":"Hello,How are you"})
```
