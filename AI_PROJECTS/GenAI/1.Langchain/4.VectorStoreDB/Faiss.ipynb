```python
from langchain_community.document_loaders import TextLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_community.embeddings import OllamaEmbeddings
from langchain_community.vectorstores import FAISS
```

```python
#data load
loader = TextLoader("speech.txt")
text_documents = loader.load()
# text_documents

# data transform
text_splitter = RecursiveCharacterTextSplitter(chunk_size = 500, chunk_overlap = 30)
final_docs = text_splitter.split_documents(text_documents)
final_docs
```

```python
#embedding 
## Bash cmd - ollama pull nomic-embed-text =>great for RAG pipelines
embeddings = OllamaEmbeddings(model="nomic-embed-text")
```

```python
#vector store with doc embeddings

db = FAISS.from_documents(final_docs,embeddings)
db
```

```python
#querying
query = "What is the speaker's attitude toward the German people?"
result = db.similarity_search(query)
result[0].page_content
```

```python
#RETREIVER

retriever = db.as_retriever()
retriever.invoke(query)
```

```python
##Similarity earch with scores

result_score = db.similarity_search_with_score(query)
result_score
```

```python
##can retreive response by querying as vectors 
query_vectors = embeddings.embed_query(query)
query_vectors
```

```python
result_of_vectors = db.similarity_search_by_vector(query_vectors)
result_of_vectors
```

```python
##saving and loading
db.save_local("faiss.index")
```

```python
new_db = FAISS.load_local("faiss.index",embeddings,allow_dangerous_deserialization=True)
docs = new_db.similarity_search(query)
```
