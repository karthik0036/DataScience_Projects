```python
# Data ingestion -  Text loader
from langchain_community.document_loaders import TextLoader

loader = TextLoader("speech.txt")
text_documents = loader.load()
text_documents
```

```python
#Data Transform - data to chunks of documents
from langchain_text_splitters import RecursiveCharacterTextSplitter
text_splitter = RecursiveCharacterTextSplitter(chunk_size = 500, chunk_overlap = 50)
final_docs = text_splitter.split_documents(text_documents)
final_docs
```

```python
from langchain_openai import OpenAIEmbeddings
embeddings = OpenAIEmbeddings(model="text-embedding-3-large")
embeddings
```

```python
# vector embedding and vector db 
from langchain_community.vectorstores import Chroma

db = Chroma.from_documents(final_docs,embeddings) #since embedding tech is openai and not free =>error
db
```

```python
#to retreive the dat from db

query = "It will be all the easier for us to conduct ourselves as belligerents"
result = db.similarity_search(query)
result
```
