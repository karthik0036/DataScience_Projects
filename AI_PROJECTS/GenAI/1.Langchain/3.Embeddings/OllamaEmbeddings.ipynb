```python
from langchain_community.embeddings import OllamaEmbeddings
```

```python
embeddings = OllamaEmbeddings( model="gemma:2b") ## by default it uses lamma2
```

```python
embeddings
```

```python
q1 = embeddings.embed_documents([
    "Apple is a healthy fruit",
    "Apple is a big phone company"
])
```

```python
len(q1[0]) # gemma mdoel creates vectors of dim 22048
```

```python
q1[1]
```

```python
q2 = embeddings.embed_query("which is the big phone company?")
```

```python
q2[0]
```
