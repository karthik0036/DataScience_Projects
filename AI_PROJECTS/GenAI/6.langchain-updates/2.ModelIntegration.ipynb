```python
import os
from dotenv import load_dotenv
load_dotenv()

os.environ["OPENAI_API_KEY"] == os.getenv("OPENAI_API_KEY")
groq_api_key = os.getenv("GROQ_API_KEY")
google_api_key = os.getenv("GOOGLE_API_KEY")
```

```python
from langchain.chat_models import init_chat_model

model = init_chat_model("gemma:2b",model_provider="ollama")
model.invoke("what is llm")
```

```python
from langchain_ollama import ChatOllama
from langchain_groq import ChatGroq
from langchain_google_genai import ChatGoogleGenerativeAI 
```

```python
llm_ollama = ChatOllama(model="gemma:2b")
llm_ChatGroq = ChatGroq(model="llama-3.1-8b-instant",groq_api_key=groq_api_key)
llm_google_genai = ChatGoogleGenerativeAI(model="gemini-flash-lite-latest",google_api_key=google_api_key)
```

```python
llm_ollama.invoke("hello,how are you?")
```

```python
llm_ChatGroq.invoke("hello")
```

```python
llm_google_genai.invoke("hey,my name is jeef")
```

```python
# Gemini Flash Lite
# Gemini 2.0 Flash Lite
# Gemma 3 1B IT

import google.generativeai as genai

genai.configure(api_key=google_api_key)

for m in genai.list_models():
    print(m.name)
```

STREAMING

```python
llm_ChatGroq.invoke("why parrot have feathers?")
```

```python
for i in llm_ChatGroq.stream("why parrots have colorful feathers?"):
    print(i.text, end ="|",flush =True)
```

BATCH

```python
responses = llm_ChatGroq.batch([
    "whats the latest ferrari car?",
    "whats the latest audi car?",
    "whats the latest dodge car?",
],
config = { "max_concurrency":5 } #limit to 5 parallel calls
                               )

for i in responses:
    print(i)
```
