```python
import os
from dotenv import load_dotenv 
load_dotenv()
```

```python
os.environ["OPENAI_API_KEY"]=os.getenv("OPENAI_API_KEY")
os.environ["LANGCHAIN_API_KEY"]=os.getenv("LANGCHAIN_API_KEY")
os.environ["LANGCHAIN_PROJECT"]=os.getenv("LANGCHAIN_PROJECT")
os.environ["LANGCHAIN_TRACING_V2"]="true"
```

```python
#Creating a simple chatbot 

#loading the llm model
from langchain_openai import ChatOpenAI
llm = ChatOpenAI(model="gpt-4o")
llm
```

```python
###input and get response since openai api key is init from llm
llm.invoke("hello, whats up?")
```

```python
from langchain_core.prompts import ChatPromptTemplate

prompt = ChatPromptTemplate.from_messages(
    [
        ("system","you are an expert ai engineer. provide the answers for user based on the questions"),
        ("user","{input}")
    ]
)
prompt
```

```python
chain = prompt | llm
response = chain.invoke({"input":"why langsmith is used?"})
response
```

```python
from langchain_core.output_parsers import StrOutputParser

parser = StrOutputParser()

chain = prompt | llm | parser
```
